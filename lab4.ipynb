# Practical 4 - Image Enhancement

Aim: Enhance an image using log transform, gamma transform and intensity stretching.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Name: Navjot Singh")
print("Roll No: 2025010071")

img = cv2.imread("images/navjot_a.jpg", cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(8, 5))
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# log transform
c = 1
log_img = c * np.log1p(img.astype(np.float32))
log_img = 255 * log_img / log_img.max()
log_img = log_img.astype(np.uint8)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(log_img, cmap="gray")
plt.title("Log Transformed Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# gamma transform for two values
gamma1 = 0.5
gamma2 = 2.0

gamma_img1 = np.uint8(255 * np.power(img / 255.0, gamma1))
gamma_img2 = np.uint8(255 * np.power(img / 255.0, gamma2))

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(gamma_img1, cmap="gray")
plt.title("Gamma = 0.5")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(gamma_img2, cmap="gray")
plt.title("Gamma = 2.0")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# stretch 80-120 to 50-150
r1, r2 = 80, 120
s1, s2 = 50, 150

print("Input range:", r1, "-", r2)
print("Output range:", s1, "-", s2)

stretched = s1 + ((s2 - s1) / (r2 - r1)) * (img.astype(np.float32) - r1)
stretched = np.uint8(np.clip(stretched, 0, 255))

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(stretched, cmap="gray")
plt.title("Intensity Stretched Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
