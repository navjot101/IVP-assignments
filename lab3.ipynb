# Practical 3 - Border and Complement

Aim: Add a border to grayscale and binary images and find binary and grayscale complements.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Name: Navjot Singh")
print("Roll No: 2025010071")

img = cv2.imread("images/navjot_a.jpg", cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(8, 5))
plt.imshow(img, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
# gray border
width = 30
color = 200
rows, cols = img.shape

padded = np.full((rows + 2 * width, cols + 2 * width), color, dtype=np.uint8)
padded[width:width + rows, width:width + cols] = img

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(padded, cmap="gray")
plt.title("Image with Border")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# binary image and its border
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
binary_width = 20
binary_color = 255
rows, cols = binary.shape

padded_binary = np.full((rows + 2 * binary_width, cols + 2 * binary_width), binary_color, dtype=np.uint8)
padded_binary[binary_width:binary_width + rows, binary_width:binary_width + cols] = binary

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(padded_binary, cmap="gray")
plt.title("Binary Image with Border")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# complements of the binary and gray images
binary_01 = (binary > 0).astype(np.uint8)
complement_binary = 1 - binary_01
complement_gray = 255 - img

plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(complement_binary * 255, cmap="gray")
plt.title("Binary Complement")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(img, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(complement_gray, cmap="gray")
plt.title("Grayscale Complement")
plt.axis("off")

plt.tight_layout()
plt.show()
```
