# Practical 2 - Binary and Grayscale Conversion

Aim: Convert a grayscale image to binary using mean and user-defined thresholds, and convert an RGB image to grayscale using average and weighted methods.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Name: Navjot Singh")
print("Roll No: 2025010071")

img = cv2.imread("images/navjot_a.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8, 5))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
# use the mean value as threshold
mean_value = gray.mean()
_, binary_mean = cv2.threshold(gray, mean_value, 255, cv2.THRESH_BINARY)
cv2.imwrite("output/lab2_binary_mean.jpg", binary_mean)
print("Mean threshold value:", round(mean_value, 2))

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary_mean, cmap="gray")
plt.title("Binary Image (mean threshold)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# try a fixed threshold
user_threshold = 100
_, binary_user = cv2.threshold(gray, user_threshold, 255, cv2.THRESH_BINARY)
cv2.imwrite("output/lab2_binary_user.jpg", binary_user)
print("User threshold value:", user_threshold)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary_user, cmap="gray")
plt.title("Binary Image (threshold = 100)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# average of the three color channels
b, g, r = cv2.split(img)
gray_average = ((b.astype(np.float32) + g.astype(np.float32) + r.astype(np.float32)) / 3).astype(np.uint8)
cv2.imwrite("output/lab2_gray_average.jpg", gray_average)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_average, cmap="gray")
plt.title("Grayscale (average method)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# weighted grayscale
wR = 0.299
wG = 0.587
wB = 0.114

gray_weighted = (wR * r + wG * g + wB * b).astype(np.uint8)
cv2.imwrite("output/lab2_gray_weighted.jpg", gray_weighted)

print("Weights:", wR, wG, wB)
print("Sum of weights:", wR + wG + wB)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(gray_average, cmap="gray")
plt.title("Average Method")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_weighted, cmap="gray")
plt.title("Weighted Method")
plt.axis("off")

plt.tight_layout()
plt.show()
```
